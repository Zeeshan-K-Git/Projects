```python
from scipy.stats import norm
import numpy as np
import pandas as pd
import yfinance as yf
```

```python
data=pd.DataFrame()
data= yf.download('^NSEI', start='2025-10-1')['Close']
data
```

```python
log_returns= np.log(1+ data.pct_change())
log_returns
```

```python
stdev= log_returns.std()*250**0.5
stdev
```

```python
def d1(s,k,r,t,stdev):
    return (np.log(s/k) + (r+ stdev**2*0.5)*t)/(stdev*(t**0.5))
```

```python
def d2(s,k,r,t,stdev):
    return (np.log(s/k) + (r- stdev**2*0.5)*t)/(stdev*(t**0.5))
```

```python
def BSM(s,k,r,t,stdev):
    return (s*norm.cdf(d1(s,k,r,t,stdev))) - (k*np.exp(-r*t)*norm.cdf(d2(s,k,r,t,stdev)))
```
